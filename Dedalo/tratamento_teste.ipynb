```python
import pandas as pd
import numpy as np 
```

```python
data = pd.read_csv("dadoslimpos.csv")
df = data.drop(columns='Pressao')

# 2. Renomeia as colunas para o padrão em inglês que o seu RocketPy estava esperando no erro anterior
motor_data = df.rename(columns={'Tempo': 'Time(s)', 'Empuxo': 'Thrust(N)'})
motor_data = motor_data.reset_index(drop=True)

motor_data['Time(s)'] = (motor_data['Time(s)'] - motor_data['Time(s)'].iloc[0]) / 1000
motor_data['Thrust(N)'] = (motor_data['Thrust(N)']/100)
```
