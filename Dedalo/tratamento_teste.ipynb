```python
import pandas as pd
import numpy as np 
```

```python
data = pd.read_csv("motor_teste_estatico_1.csv")
```

```python
df = data.drop(columns='Pressao')
```

```python
inicio = df[df['Tempo'] == 165281].index[0]
```

```python
fim = df[df['Tempo'] == 173851].index[0]
```

```python
motor_data = df.loc[inicio:fim]
```

```python
motor_data
```

```python
motor_data['Tempo'] = (motor_data['Tempo'] - motor_data['Tempo'].iloc[0]) / 1000
motor_data['Empuxo'] = (motor_data['Empuxo']/1000)
```

```python
# 2. Renomeia as colunas para o padrão em inglês que o seu RocketPy estava esperando no erro anterior
motor_data = motor_data.rename(columns={'Tempo': 'Time(s)', 'Empuxo': 'Thrust(N)'})
```

```python
motor_data.to_csv('motor_data_pronto.csv', index=False)
```
